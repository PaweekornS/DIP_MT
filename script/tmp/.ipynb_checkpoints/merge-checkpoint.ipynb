```python
from transformers import AutoTokenizer, BitsAndBytesConfig
from peft import PeftModel
import torch
import glob

root_dir = "/project/lt200304-dipmt/paweekorn/models"
compute_dtype = torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)
```

## CausalLM

```python
from transformers import AutoModelForCausalLM

model_id = "gemma3-4b-pt"
base_model = AutoModelForCausalLM.from_pretrained(
    f"{root_dir}/base/{model_id}",
)
tokenizer = AutoTokenizer.from_pretrained(f"{root_dir}/base/{model_id}")

merged_model = PeftModel.from_pretrained(base_model, f"{root_dir}/adapter/{model_id}/checkpoint-1242")
merged_model = merged_model.merge_and_unload()

merged_model.save_pretrained(f"{root_dir}/fine-tuned/{model_id}", safe_serialization=True)
tokenizer.save_pretrained(f"{root_dir}/fine-tuned/{model_id}")
```

## Gemma 3 Series

```python
from transformers import AutoProcessor, Gemma3ForConditionalGeneration

MODEL_ID = "gemma3-4b-pt"
MODEL_PATH = f"{root_dir}/base/{MODEL_ID}"
ADAPTER_PATH = f"{root_dir}/adapter/{MODEL_ID}/checkpoint-1242"
output_path = f"{root_dir}/fine-tuned/{MODEL_ID}"

base_model = Gemma3ForConditionalGeneration.from_pretrained(
    MODEL_PATH, 
    tie_word_embeddings=False,
)
processor = AutoProcessor.from_pretrained(MODEL_PATH, use_fast=False)

merged_model = PeftModel.from_pretrained(base_model, ADAPTER_PATH)
merged_model = merged_model.merge_and_unload()

merged_model.save_pretrained(output_path, safe_serialization=True)
processor.save_pretrained(output_path)
print("Save merged model at:", output_path)
```

## NLLB

```python
from transformers import M2M100ForConditionalGeneration

BASE_ID     = f"{root_dir}/base/nllb-3.3b"
ADAPTER_DIR = f"{root_dir}/adapter/nllb-3.3b/checkpoint-1275"
OUT_DIR     = f"{root_dir}/fine-tuned/nllb-3.3b"          

base = M2M100ForConditionalGeneration.from_pretrained(BASE_ID, torch_dtype=torch.bfloat16,
                                                     quantization_config=bnb_config)

# Attach adapters
peft_model = PeftModel.from_pretrained(base, ADAPTER_DIR)
merged = peft_model.merge_and_unload()

# Save merged model + tokenizer (safetensors)
merged.save_pretrained(OUT_DIR, safe_serialization=True)
tok = AutoTokenizer.from_pretrained(BASE_ID, use_fast=True)
tok.save_pretrained(OUT_DIR)

print(f"✅ Merged model saved to: {OUT_DIR}")
```

```python
import pandas as pd
import numpy as np

from tqdm import tqdm
import json
import re

def check_bracket_consistency(eng_text, tha_text):
    eng_brackets = re.findall(r'[^a-zA-Z]', eng_text)
    tha_brackets = re.findall(r'[^\u0E00-\u0E7F]', tha_text)

    eng_brackets = len([x for x in eng_brackets if not x.isspace()])
    tha_brackets = len([x for x in tha_brackets if not x.isspace()])

    return eng_brackets == tha_brackets

def clean_thai_spacing(text):
    filtered = re.findall(r'[^\u0E00-\u0E7F]', text)
    if all([x.isspace() for x in filtered]):
        text = re.sub(r'\s', '', text)
    return text
```
